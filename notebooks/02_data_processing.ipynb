```python
import pandas as pd
import re
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Load combined dataset
df = pd.read_csv('../data/raw/combined_transactions.csv')

# Text cleaning
def clean_text(text: str) -> str:
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    return re.sub(r'\s+', ' ', text).strip()

df['cleaned_description'] = df['raw_description'].apply(clean_text)

# Encode targets
label_encoder = LabelEncoder()
df['label'] = label_encoder.fit_transform(df['vat_code'])

# Save encoder artifact
joblib.dump(label_encoder, '../model/artifacts/label_encoder.joblib')

# Stratified Split (70/15/15)
train_val, test_df = train_test_split(
    df, test_size=0.15, random_state=42, stratify=df['label']
)

train_df, val_df = train_test_split(
    train_val, test_size=0.1765, random_state=42, stratify=train_val['label']
)

# Export split files
train_df.to_csv('../data/processed/train.csv', index=False)
val_df.to_csv('../data/processed/val.csv', index=False)
test_df.to_csv('../data/processed/test.csv', index=False)

print(f"Data split complete: {len(train_df)} Train | {len(val_df)} Val | {len(test_df)} Test")
```
